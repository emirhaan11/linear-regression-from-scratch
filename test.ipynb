```python
import numpy as np
from model import LinearRegression
```

```python
X_train = np.random.rand(100, 2)
y_train = 3 * X_train[:, 0] + 5 * X_train[:, 1] + 10 + np.random.randn(100) * 0.1
# y = 3*x1 + 5*x2 + 10 + noise
# Expected: w1 ≈ 3, w2 ≈ 5, b ≈ 10

print(f"Shape -> X: {X_train.shape}, y: {y_train.shape}")
```

```python
model = LinearRegression(learning_rate=0.1, iters=2000)
model.fit(X_train, y_train)
```

```python
print("\n" + "="*40)
print("          COMPARISON OF RESULTS")
print("="*40)
print(f"{'Parameter':<15} | {'Real Value':<10} | {'Model Prediction':<15}")
print("-" * 45)
print(f"{'Bias (b)':<15} | {10:<10} | {model.bias:.5f}")
print(f"{'Weight 1 (w1)':<15} | {3:<10} | {model.weights[0]:.5f}")
print(f"{'Weight 2 (w2)':<15} | {5:<10} | {model.weights[1]:.5f}")
print("-" * 45)


test_x = np.array([[0.5, 0.5]])
pred = model.predict(test_x)
print(f"\nTest (x1=0.5, x2=0.5):")
print(f"Expected Value : 14.0")
print(f"Model Predict  : {pred[0]:.5f}")
```
